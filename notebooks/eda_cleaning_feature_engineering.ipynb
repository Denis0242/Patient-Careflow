```python
import pandas as pd
df = pd.read_csv('../data/careflow_dataset.csv')
df.head()
```

```python
df.info()
df.isna().sum()
df.duplicated().sum()
```

```python
df.columns = [c.strip().lower().replace(' ', '_').replace('-', '_') for c in df.columns]
df['journey_date'] = pd.to_datetime(df['journey_date'], errors='coerce')
df.to_csv('../data/careflow_clean.csv', index=False)
```
